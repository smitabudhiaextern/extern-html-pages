```python
# Cell 2: give spaCy some text and see what it finds.

# Change the words between the quotes, then run this cell again.
text = "Lot Number: 18356721 Date of Manufacture: 20240315"

import importlib
importlib.invalidate_caches()
import spacy

# What each spaCy label means, in plain English
MEANINGS = {
    "PERSON": "A person's name",
    "NORP": "A nationality, religion or political group",
    "FAC": "A building, airport, road or bridge",
    "ORG": "An organization: a company or a group of people",
    "GPE": "A country, city or state",
    "LOC": "A place that isn't a country or city, like a mountain or a river",
    "PRODUCT": "A thing you can buy, like a phone or a car",
    "EVENT": "A named event, like a war or a sports match",
    "WORK_OF_ART": "A title, like a book or a song",
    "LAW": "A named law or legal document",
    "LANGUAGE": "A language, like English or Spanish",
    "DATE": "A date",
    "TIME": "A time of day",
    "PERCENT": "A percentage",
    "MONEY": "An amount of money",
    "QUANTITY": "A measurement, like a weight or a distance",
    "ORDINAL": "A position in order, like first or 2nd",
    "CARDINAL": "A plain number",
}

nlp = spacy.load("en_core_web_sm")
doc = nlp(text)

print("Your text:", text)
print()
print("1. Tokens: spaCy split your text into", len(doc), "pieces")
print("   " + " | ".join(token.text for token in doc))
print()
print("2. Entities: the important pieces spaCy labeled")
if len(doc.ents) == 0:
    print("   spaCy didn't label anything in this text.")
for ent in doc.ents:
    meaning = MEANINGS.get(ent.label_, "No plain-English meaning listed")
    print(f"   {ent.text!r:32} {ent.label_:12} {meaning}")
```
